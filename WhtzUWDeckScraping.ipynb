```python
import numpy as np
import pandas as pd
import scrython
import time


#this operates essentially as a giant dataframe, and manipulation of the zone column lets you know what is where(Board, Deck, Exile, Graveyard, Hand, etc.)
#full list of deck functions is found in deckfunctions.py

#Get every blue/white card via scrython and put them into the deck dataframe
#scryfall(and consequenctly scrython) paginates their results to 175 cards per page, this code iterates through all of them and pulls relevant info for play, obtaining a deck of 14592 cards(at the time of writing)

#This notebook is to just show how the deck was put together, id just use the prepared csv if you want to play with the deck

deck = []
for i in range(1,85):
    time.sleep(5) #avoid a timeout
    results = scrython.cards.Search(q="c<=uw -c:red -c:black -c:green",page=i)
    print(f"{results.total_cards} total cards, cycle {i}")
    for card in results.data:
        row = {
            'Name' : card.name,
            'Type' : card.type_line,
            'Cost' : card.mana_cost,
            'Text' : card.oracle_text,
            'Power': card.power,
            'Toughness': card.toughness,
            'Zone' : 'Deck'
        }
        deck.append(row)
    

df = pd.DataFrame(deck)
df.shape
```

```python
df.to_csv('whtzfulluwdecklist.csv') #convert to csv so we dont need to scrape the API everytime
```
