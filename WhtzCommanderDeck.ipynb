```python
import numpy as np
import pandas as pd
import scrython
import deckfunctions


#this operates essentially as a giant dataframe, and manipulation of the zone column lets you know what is where(Board, Deck, Exile, Graveyard, Hand, etc.)
#full list of deck functions is found in deckfunctions.py

#Get every blue/white card via scrython(basic card search rn), and put them into the deck dataframe


results = scrython.cards.Search(q="c<=uw")
print(f"Found {results.total_cards} cards")
deck = []
for card in results.data:
    row = {
        'Name' : card.name,
        'Type' : card.type_line,
        'Cost' : card.mana_cost,
        'Text' : card.oracle_text,
        'Power': card.power,
        'Toughness': card.toughness,
        'Zone' : 'Deck'
    }
    deck.append(row)

#Add Lands to dataframe(50 each)
for i in range(50):
    islands = {
        'Name' : 'Island',
        'Type' : 'Land',
        'Cost' : 0,
        'Text' : 'Land',
        'Zone' : 'Deck'
    }
    plains = {
        'Name' : 'Plains',
        'Type' : 'Land',
        'Cost' : 0,
        'Zone' : 'Deck'
    }
    deck.append(islands)
    deck.append(plains)
    

df = pd.DataFrame(deck)
df.shape
```
